```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



```

```python
df = pd.read_csv("superstore_sales.csv")

print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())
```

```python
print("Data types and non-null values:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nStatistical summary:")
display(df.describe(include="all").T)
```

```python
# Check rows where the main business fields are missing
print("Rows missing Order Date:", df["Order Date"].isna().sum())
print("Rows missing Sales:", df["Sales"].isna().sum())

# Check exact duplicate rows
print("Exact duplicate rows:", df.duplicated().sum())

# Check duplicate Row IDs
print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())

# Check rows with no order date
display(df[df["Order Date"].isna()].head())
```

```python
# Keep a copy of the original data
df_original = df.copy()

# Convert date columns to datetime
df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

# Keep records with the essential fields for sales analysis
df_clean = df.dropna(
    subset=["Order Date", "Sales", "Profit", "Category", "Region", "Segment"]
).copy()

print("Original rows:", len(df_original))
print("Rows after removing incomplete business records:", len(df_clean))
print("Remaining missing values:")
print(df_clean[["Order Date", "Sales", "Profit"]].isna().sum())
```

```python
print("Exact duplicates in cleaned data:", df_clean.duplicated().sum())

print(
    "Duplicate Order IDs:",
    df_clean["Order ID"].duplicated().sum()
)
```

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()
total_quantity = df_clean["Quantity"].sum()

profit_margin = total_profit / total_sales * 100

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Profit: ${total_profit:,.2f}")
print(f"Unique Orders: {total_orders:,}")
print(f"Total Quantity Sold: {total_quantity:,.0f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

```python
category_summary = (
    df.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
)

print(category_summary)
```

```python
region_summary = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Total_Profit", ascending=False)
)

print(region_summary)
```

```python
category_sales = (
    df_clean.groupby("Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
)

category_sales["Profit_Margin_%"] = (
    category_sales["Total_Profit"]
    / category_sales["Total_Sales"] * 100
)

display(category_sales.round(2))
```

```python
region_summary = (
    df.groupby("Region")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .reset_index()
    .sort_values("Total_Profit", ascending=False)
)

print(region_summary)
```

```python
segment_summary = (
    df.groupby("Segment")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum"),
        Total_Quantity=("Quantity", "sum")
    )
    .sort_values("Total_Sales", ascending=False)
    .reset_index()
)

print(segment_summary)
```

```python
# Extract Year-Month from Order Date
df["Year-Month"] = df["Order Date"].dt.to_period("M")

# Calculate total sales for each month
monthly_sales = (
    df.groupby("Year-Month")["Sales"]
    .sum()
    .reset_index()
)

# Display the results
print(monthly_sales)
```

```python
monthly_sales["Year-Month"] = (
    monthly_sales["Year-Month"].astype(str)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(14, 6))

sns.lineplot(
    data=monthly_sales,
    x="Year-Month",
    y="Sales",
    marker="o"
)

plt.title("Monthly Sales Trend (2015–2018)")
plt.xlabel("Year-Month")
plt.ylabel("Total Sales ($)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df_clean,
    x="Discount",
    y="Profit",
    alpha=0.6
)

plt.axhline(y=0, linestyle="--")
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit ($)")
plt.tight_layout()
plt.show()
```

## Question 1: Which product sub-category generates the highest total profit?

```python

subcategory_summary = (
    df_clean.groupby("Sub-Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values("Total_Profit", ascending=False)
)

print("Highest-profit subcategories:")
print(subcategory_summary.head())

print("\nLowest-profit subcategories:")
print(subcategory_summary.tail())
```

### Question 2: Which product sub-category has the lowest total sales, and what factors might explain its performance?

```python

lowest_sales = (
    df_clean.groupby("Sub-Category")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values("Total_Sales", ascending=True)
)

print(lowest_sales.head())
```

### Question 3: Which year generated the highest total profit, and what factors may have contributed to it?

```python

df_clean["Year"] = df_clean["Order Date"].dt.year

yearly_profit = (
    df_clean.groupby("Year")
    .agg(
        Total_Sales=("Sales", "sum"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
    .sort_values("Total_Profit", ascending=False)
)

print(yearly_profit)
```

### Question 5: Which month recorded the highest and lowest total sales across the entire dataset?

```python

highest_month = monthly_sales.loc[monthly_sales["Sales"].idxmax()]
lowest_month = monthly_sales.loc[monthly_sales["Sales"].idxmin()]

print("Month with Highest Sales:")
print(highest_month)

print("\nMonth with Lowest Sales:")
print(lowest_month)
```

3. **Tables Require Profitability Improvement:** Tables recorded the largest sub-category loss of $17,725.48. The business should investigate product costs, discount levels, and pricing strategies to reduce these losses.

4. **The West Region Performs Best:** The West region generated the highest regional profit of $108,418.45, while the Central region generated the lowest at $39,706.36. The company should investigate the differences in product mix, discounting, and sales performance between these regions.

5. **Discounts and Profitability Need Careful Monitoring:** The discount-versus-profit scatter plot suggests that higher discounts are associated with more loss-making transactions. The company should review high-discount sales and establish discount limits that protect profit margins.

### Overall Recommendation

The business should focus on profitable products, review loss-making sub-categories, investigate regional performance differences, and manage discounts carefully. Regular monitoring of sales, profit, and profit margins can help improve business decisions.
